```python
import sys; sys.path.insert(0,"../src")
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from market_data import load_returns
from factor_analysis import compute_beta_alpha, rolling_beta, rolling_correlations, annual_factor_decomp
returns = load_returns("../data/market_returns.csv")
weights = np.array([0.35,0.25,0.20,0.10,0.10])
port_ret = (returns * weights).sum(axis=1)
spy_ret  = returns["SPY"]
print("Data loaded")
```
