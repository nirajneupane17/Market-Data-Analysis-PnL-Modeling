```python
import sys; sys.path.insert(0,"../src")
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from market_data import load_returns, compute_drawdown, rolling_metrics
from pnl_engine import compute_portfolio_pnl
returns = load_returns("../data/market_returns.csv")
weights = np.array([0.35,0.25,0.20,0.10,0.10])
port_ret = (returns * weights).sum(axis=1)
dd = compute_drawdown(port_ret)
print("Data loaded")
```

```python
sharpe  = port_ret.mean()/port_ret.std()*np.sqrt(252)
sortino = port_ret.mean()/port_ret[port_ret<0].std()*np.sqrt(252)
calmar  = port_ret.mean()*252/abs(dd.min())
print(f"Sharpe Ratio  : {sharpe:.4f}")
print(f"Sortino Ratio : {sortino:.4f}")
print(f"Calmar Ratio  : {calmar:.4f}")
print(f"Max Drawdown  : {dd.min()*100:.2f}%")
print(f"Ann. Return   : {port_ret.mean()*252*100:.2f}%")
print(f"Ann. Vol      : {port_ret.std()*np.sqrt(252)*100:.2f}%")
```
