```python
import sys; sys.path.insert(0,"../src")
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from market_data import load_returns
from pnl_engine import compute_portfolio_pnl, pnl_attribution, annual_pnl, monthly_pnl
print("Libraries loaded")
```

```python
returns = load_returns("../data/market_returns.csv")
weights = np.array([0.35,0.25,0.20,0.10,0.10])
assets = list(returns.columns)
pnl_df = compute_portfolio_pnl(returns, weights, notional=1_000_000)
print(f"Portfolio PnL shape: {pnl_df.shape}")
print(f"Total PnL: USD {pnl_df['portfolio_pnl'].sum():,.0f}")
print(f"Cumulative PnL: USD {pnl_df['cumulative_pnl'].iloc[-1]:,.0f}")
```

```python
attr = pnl_attribution(pnl_df, assets)
print("PnL Attribution:")
for k,v in attr.items(): print(f"  {k}: {v}")
```

```python
print("Annual PnL:")
print(annual_pnl(pnl_df).to_string())
```
